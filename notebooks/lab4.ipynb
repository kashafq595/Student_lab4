## Part 10: Matrix operations and linear algebra

```python
# Example
A = np.array([[1, 2], [3, 4]])
B = np.array([[10, 20], [30, 40]])
print(A * B)
print(A @ B)
print(A.T)
X = np.array([[1, 2, 3], [4, 5, 6]])
print((X @ X.T).shape)

M = np.array([[2, 1], [1, 3]])
rhs = np.array([5, 10])
sol = np.linalg.solve(M, rhs)
print(sol, np.linalg.det(M), np.allclose(M @ sol, rhs))
print(np.linalg.inv(M))
print(np.linalg.norm(np.array([3, 4])))

# Try it yourself 10
# 1. shapes
print((np.ones((4, 3)) @ np.ones((3, 5))).shape)
try:
    np.ones((4, 3)) @ np.ones((4, 3))
except ValueError as e:
    print('Error:', e)

# 2. weighted average
marks3 = np.array([[80, 70, 90], [60, 50, 40]])
weights = np.array([0.5, 0.3, 0.2])
print(marks3 @ weights)

# 3. solve 3x + 2y = 12 and x - y = 1
M2 = np.array([[3, 2], [1, -1]])
rhs2 = np.array([12, 1])
sol2 = np.linalg.solve(M2, rhs2)
print(sol2, np.allclose(M2 @ sol2, rhs2))

# 4. dot product
print(np.dot([1, 2, 3], [4, 5, 6]))
```

## Part 12: Performance measurement: loop vs vectorized

```python
import time, timeit, tracemalloc

N = 1_000_000
py_list = list(range(N))
np_arr = np.arange(N)

def loop_double(values):
    result = []
    for x in values:
        result.append(x * 2)
    return result

def numpy_double(values):
    return values * 2

for label, func, data in (('loop', loop_double, py_list), ('numpy', numpy_double, np_arr)):
    start = time.perf_counter()
    func(data)
    print(label, round(time.perf_counter() - start, 5), 'seconds')
```

```python
n = 20
t_loop = timeit.timeit('loop_double(py_list)', globals=globals(), number=n) / n
t_np = timeit.timeit('numpy_double(np_arr)', globals=globals(), number=n) / n
print('loop', round(t_loop, 5), 'numpy', round(t_np, 5), 'ratio', round(t_loop / t_np))
```

```python
for label, make in (('list', lambda: [x * 2 for x in range(N)]),
                    ('array', lambda: np.arange(N) * 2)):
    tracemalloc.start()
    obj = make()
    current, peak = tracemalloc.get_traced_memory()
    tracemalloc.stop()
    print(label, 'current MB', round(current / 1e6, 1), 'peak MB', round(peak / 1e6, 1))
```

```python
rows = []
for size, number in [(1_000, 100), (100_000, 100), (1_000_000, 20)]:
    lst = list(range(size))
    arr = np.arange(size)
    t_loop = timeit.timeit('loop_double(lst)',
                           globals={'loop_double': loop_double, 'lst': lst},
                           number=number) / number
    t_np = timeit.timeit('numpy_double(arr)',
                         globals={'numpy_double': numpy_double, 'arr': arr},
                         number=number) / number
    rows.append((size, t_loop, t_np, round(t_loop / t_np)))

print(f"{'N':>10} {'t_loop (s)':>12} {'t_np (s)':>12} {'ratio':>8}")
for size, tl, tn, ratio in rows:
    print(f"{size:>10} {tl:>12.6f} {tn:>12.6f} {ratio:>8}")
```

### Part 12 conclusion (size sweep)

With my numbers, the ratio was 40 at N = 1,000, 112 at N = 100,000 and 29 at N = 1,000,000, so the ratio did not keep growing with N. In all three sizes NumPy was much faster than the loop (about 29 to 112 times), and the loop time grew as N grew (0.000227 s to 0.140064 s).

### Why is the best of several runs fairer?

Other programs running on my computer can only make my code slower, never faster. So the smallest time out of several runs is the one least disturbed by other things, and it is closest to the true speed of the code.

```python
n = 20
best_loop = min(timeit.repeat('loop_double(py_list)', globals=globals(), number=n, repeat=5)) / n
best_np = min(timeit.repeat('numpy_double(np_arr)', globals=globals(), number=n, repeat=5)) / n
print('best loop', round(best_loop, 5), 'best numpy', round(best_np, 5), 'ratio', round(best_loop / best_np))
```
