```python
import numpy as np
scores = np.array([[45, 78, 90, 60], 
                   [32, 65, 70, 55], 
                   [88, 92, 79, 95]]) # 3 students x 4 subjects 
print(scores.ndim, scores.shape, scores.size, scores.dtype) 
print(np.zeros((2, 3))) 
print(np.arange(0, 10, 2), np.linspace(0, 1, 5)) 
lst = [1, 2, 3] 
arr = np.array(lst) 
print(lst * 2, arr * 2)
```

   ## Part 5: Arrays, ndim, shape, dtype

```python
print(np.ones((2, 3)))
print(np.full((2, 2), 7))
print(np.arange(2, 11, 2))

z = np.zeros((2, 3, 4))
print(z.ndim, z.shape, z.size)

print(np.array([1, 2.5, 3]).dtype)
```

## Part 6: Indexing and slicing

```python
# Example
print(scores[0, 1])
print(scores[-1, -2])
print(scores[:, 0])
print(scores[0:2, 1:3])
view = scores[0, :2]
view[0] = 0
print(scores[0])
scores[0, 0] = 45
copy = scores[1, :2].copy()
copy[0] = 999
print(scores[1])

# Try it yourself 6
print(scores[-1, :2])
print(scores[0:2, 2])
print(scores[0:2, 2:4])

v = scores[0, :2]
v[0] = 0
print("after slice edit:", scores[0])
scores[0, 0] = 45
c = scores[0, :2].copy()
c[0] = 0
print("after copy edit:", scores[0])
```

## Part 7: Reshape

```python
# Example
a = np.arange(12)
print(a.reshape(3, 4))
print(a.reshape(2, -1).shape)
print(np.array([1, 2, 3]).reshape(3, 1))
try:
    a.reshape(5, 3)
except ValueError as e:
    print('Error:', e)

# Try it yourself 7
for s in [(4, 3), (5, 2), (2, 2, 3), (6, -1)]:
    try:
        print(s, '->', a.reshape(s).shape)
    except ValueError as e:
        print(s, 'Error:', e)

col = np.arange(6).reshape(6, 1)
print(col.shape)
print(scores.flatten())
```

## Part 8: Broadcasting

```python
# Example
bonus = np.array([5, 0, 10, 0])
print(scores + bonus)
col = np.array([[1], [2], [3]])
print(col + np.array([10, 20, 30]))
means = scores.mean(axis=0)
print(np.round(scores - means, 1))

# Try it yourself 8
for s1, s2 in [((4, 3), (3,)), ((4, 3), (4,)), ((4, 1), (1, 5))]:
    try:
        print(s1, s2, '->', (np.ones(s1) + np.ones(s2)).shape)
    except ValueError as e:
        print(s1, s2, 'Error:', e)

student_bonus = np.array([0, 5, 10]).reshape(3, 1)
print(scores + student_bonus)

print(np.round(scores - scores.mean(axis=1, keepdims=True), 1))
```
